## Functional dependency
The functional dependency that we will focus on is
- album, year -> genre

The function below determines all violations of this functional dependency. Remember that this functional dependency means that album and year should determine the value for genre, i.e., for every two rows with the same value of album and year, the value of genre should also be the same.

### Violations in our example table
Notice that there are multiple violations in the table for this functional dependency. Also notice that the song "Beat It" is involved in two violations.

```python
def fd_violations(df):
    """
    Determines all violations of the functional dependency album, year -> genre

    Args:
        df: DataFrame to check for violations

    Returns:
        List of violation pairs (row indices)
    """
    violations = []

    # Loop over all pairs of rows
    for i in range(len(df)):
        for j in range(i + 1, len(df)):
            row1 = df.iloc[i]
            row2 = df.iloc[j]

            # Check if album and year are the same
            if row1['album'] == row2['album'] and row1['year'] == row2['year']:
                # Check if genre is different (violation)
                if row1['genre'] != row2['genre']:
                    violations.append((i, j))

    return violations

def display_violation(df, row1_idx, row2_idx, violation_number=None):
    """
    Displays the details of a functional dependency violation between two rows.

    Args:
        df: DataFrame containing the data
        row1_idx: Index of the first row in violation
        row2_idx: Index of the second row in violation
        violation_number: Optional violation number for display
    """
    row1 = df.iloc[row1_idx]
    row2 = df.iloc[row2_idx]

    if violation_number:
        print(f"Violation {violation_number}:")
    else:
        print(f"Violation (rows {row1_idx} and {row2_idx}):")

    print(f"  Row {row1_idx}: {row1['song_title']} - Album: {row1['album']}, Year: {row1['year']}, Genre: {row1['genre']}")
    print(f"  Row {row2_idx}: {row2['song_title']} - Album: {row2['album']}, Year: {row2['year']}, Genre: {row2['genre']}")
    print()

# Execute the function and display results
violations = fd_violations(data)

print(f"Functional dependency violations for 'album, year -> genre':")
print(f"Number of violations found: {len(violations)}")

if violations:
    print("\nViolation details:")
    for i, (row1_idx, row2_idx) in enumerate(violations):
        display_violation(data, row1_idx, row2_idx, i+1)
else:
    print("No violations found - the functional dependency holds for this dataset.")
```

## Generating possible repairs

When a functional dependency violation is detected between two rows, there are exactly four possible ways to repair the violation for the FD **album, year → genre**:

1. **Delete first row** - Remove the first conflicting row entirely from the dataset
2. **Delete second row** - Remove the second conflicting row entirely from the dataset  
3. **Update first row's genre** - Change the genre of the first row to match the second row's genre
4. **Update second row's genre** - Change the genre of the second row to match the first row's genre

### Why these repairs work:
- **Deletion repairs** eliminate one of the conflicting rows, so no violation remains
- **Update repairs** make both rows have the same genre value for the same album+year combination, satisfying the functional dependency

Each repair has different implications:
- Deletions result in data loss but guarantee no further conflicts for that pair
- Updates preserve all data but may introduce new conflicts if the updated genre conflicts with other rows sharing the same album+year combination

## Generating all possible repaired tables

To systematically explore all ways to repair the functional dependency violations, we need to generate **all combinations** of repair choices across all violations:

### Process overview:
1. **Identify all violations** - Find every pair of rows that violates the FD
2. **Generate repairs per violation** - For each violation, create the 4 possible repairs
3. **Create repair combinations** - Generate all possible ways to choose one repair per violation
4. **Clean repair combinations** - Remove duplicates and conflicting operations
5. **Calculate repair costs** - Determine cost for each repair combination (updates=1, deletes=3)
6. **Apply repair combinations** - Apply each cleaned combination to create a repaired table
7. **Validate results** - Check which repaired tables actually satisfy the FD
8. **Find cheapest repairs** - Identify minimum cost and filter to only cheapest valid tables

### Rationale for cost-based repair selection:
We assume that **the cheapest repair is most likely to be the correct repair**. The intuition is that ordinarily the data would satisfy the functional dependency, hence the violation is caused by an **accident**: either someone accidentally adds a row for a song that was already there and the added row has slightly different data, or someone accidentally enters a wrong value for an attribute. 

The possible repairs strategy tries to **reverse the accidents**. Not knowing which accidents may have occurred, we enumerate all possibilities. Since some accidents are more likely to occur than others, we use the notion of **cost of a repair**. The repaired table with the least cost represents the situation with the **most likely accidents**, hence is most likely to be the correctly repaired table (or tables if there is more than one with the same least cost).

### Combinatorial explosion:
- With **n violations**, each having **4 repair options**, we get **4ⁿ possible combinations**
- For example: 4 violations → 4⁴ = **256 possible repaired tables**
- Each combination represents a different "repair strategy" for the entire dataset

### Key challenges:
- **Order matters for deletions** - Deleting row indices must be done in reverse order to avoid index shifting
- **Repair interactions** - Some repair combinations may introduce new violations
- **Duplicate repairs** - Same repair operation may appear multiple times in a combination
- **Conflicting repairs** - Updates to rows that are also being deleted are meaningless
- **Cost optimization** - Among valid repairs, finding the most cost-effective solutions
- **Validation required** - Not all repair combinations guarantee FD satisfaction

### Cost model:
- **Update operations**: Cost = 1 (minimal data modification - fixing typos/errors)
- **Delete operations**: Cost = 3 (data loss is more expensive - removing duplicate entries)
- **Total cost**: Sum of all operations in a repair combination

### Implementation approach:
- Use `itertools.product()` to generate all combinations efficiently
- Apply repairs in sorted order (deletions first, in reverse index order)
- **Clean repair sets** using `analyse_repair_set_validity()` to:
  - Remove duplicate repair operations (same operation, row, and arguments)
  - Remove updates to rows that are being deleted (conflicting operations)
- **Calculate costs** using `cost_of_set_of_repairs()` for each repair combination
- **Filter for validity** using `check_fd_satisfaction()` to ensure FD compliance
- **Optimize for cost** using `determine_cheapest_repairs()` to find minimum-cost solutions

Function `determine_cheapest_repairs` needs to the lowest cost. The exercise is to fill in this bit of code that iterates over all repaired tables in search for the ones with the lowest cost. Store this cost in the `min_cost variable`.

```python
def apply_single_repair(df, repair):
    """
    Applies a single repair operation to a dataframe.

    Args:
        df: DataFrame to repair
        repair: Dictionary containing repair operation details

    Returns:
        New DataFrame with the repair applied
    """
    df_repaired = df.copy()

    if repair['operation'] == 'delete':
        df_repaired = df_repaired.drop(repair['row_index']).reset_index(drop=True)
    elif repair['operation'] == 'update':
        df_repaired.loc[repair['row_index'], repair['attribute']] = repair['new_value']

    return df_repaired

def check_fd_satisfaction(df):
    """
    Checks if a dataframe satisfies the functional dependency album, year -> genre.

    Args:
        df: DataFrame to check

    Returns:
        Boolean indicating if FD is satisfied (True) or violated (False)
    """
    violations = fd_violations(df)
    return len(violations) == 0

def cost_of_set_of_repairs(repairs):
    """
    Determines the cost of a set of repairs.
    Update operations cost 1, delete operations cost 3.
    
    Args:
        repairs: List of repair dictionaries
    
    Returns:
        Integer representing the total cost
    """
    total_cost = 0
    for repair in repairs:
        if repair['operation'] == 'delete':
            total_cost += 3
        elif repair['operation'] == 'update':
            total_cost += 1
    return total_cost

def analyse_repair_set_validity(repairs):
    """
    Analyses a set of repairs and removes redundant or conflicting operations.
    - Removes duplicate repairs (same operation and arguments)
    - Removes updates to rows that are also being deleted
    
    Args:
        repairs: List of repair dictionaries
    
    Returns:
        List of cleaned repair dictionaries
    """
    cleaned_repairs = []
    seen_repairs = set()
    deleted_rows = set()
    
    # First pass: identify deleted rows and collect unique repairs
    for repair in repairs:
        # Create a unique identifier for each repair
        if repair['operation'] == 'delete':
            repair_id = ('delete', repair['row_index'])
            deleted_rows.add(repair['row_index'])
        else:  # update
            repair_id = ('update', repair['row_index'], repair['attribute'], repair['new_value'])
        
        # Only add if we haven't seen this exact repair before
        if repair_id not in seen_repairs:
            seen_repairs.add(repair_id)
            cleaned_repairs.append(repair)
    
    # Second pass: remove updates to rows that are being deleted
    final_repairs = []
    for repair in cleaned_repairs:
        if repair['operation'] == 'update' and repair['row_index'] in deleted_rows:
            # Skip this update because the row is being deleted
            continue
        final_repairs.append(repair)
    
    return final_repairs

def apply_repairs(df, violations, verbose=False):
    """
    For each violation determines possible repairs, then generates all combinations
    of repairs (one per violation) and applies them to create repaired dataframes.

    Args:
        df: Original dataframe
        violations: List of violation pairs
        verbose: If True, prints details for each repaired table

    Returns:
        List of tuples (repaired_dataframe, cost) for all possible repaired dataframes
    """
    if not violations:
        return [(df, 0)]

    # Get all possible repairs for each violation
    all_repairs = []
    for row1_idx, row2_idx in violations:
        repairs = possible_repairs(df, row1_idx, row2_idx)
        all_repairs.append(repairs)

    # Generate all combinations of repairs (one repair per violation)
    from itertools import product
    repair_combinations = list(product(*all_repairs))

    repaired_tables = []

    # Apply each combination of repairs
    for i, repair_combination in enumerate(repair_combinations):
        current_df = df.copy()

        if verbose:
            print(f"\nRepaired table {i+1}:")
            print("Repair operations:")

        # Sort repairs by operation type and row index to handle deletions properly
        # Deletions should be applied in reverse order to avoid index shifting issues
        sorted_repairs = sorted(repair_combination,
                              key=lambda r: (r['operation'] != 'delete',
                                           -r['row_index'] if r['operation'] == 'delete' else r['row_index']))
        
        # Analyse and clean the repair set
        sorted_repairs = analyse_repair_set_validity(sorted_repairs)
        
        # Calculate the cost of this repair set
        repair_cost = cost_of_set_of_repairs(sorted_repairs)

        # Apply all repairs in this combination
        for repair in sorted_repairs:
            if verbose:
                print(f"  - {repair['description']}")
            current_df = apply_single_repair(current_df, repair)

        # Check if the repaired table satisfies the functional dependency
        fd_satisfied = check_fd_satisfaction(current_df)

        if verbose:
            print(f"Repair cost: {repair_cost}")
            print(f"Functional dependency satisfied: {fd_satisfied}")

        repaired_tables.append((current_df, repair_cost))

    return repaired_tables

def filter_valid_repairs(repaired_tables_with_cost):
    """
    Filters repaired tables to only include those that satisfy the functional dependency.
    Preserves cost information.

    Args:
        repaired_tables_with_cost: List of tuples (dataframe, cost)

    Returns:
        List of tuples (dataframe, cost) that satisfy the functional dependency
    """
    valid_tables = []
    for table, cost in repaired_tables_with_cost:
        if check_fd_satisfaction(table):
            valid_tables.append((table, cost))
    return valid_tables

def determine_cheapest_repairs(repaired_tables_with_cost):
    """
    Determines the least cost among all repaired tables and returns only those
    tables that have this minimum cost.
    
    Args:
        repaired_tables_with_cost: List of tuples (dataframe, cost)
    
    Returns:
        List of tuples (dataframe, cost) with minimum cost
    """
    if not repaired_tables_with_cost:
        return []
    
    # Find the minimum cost

    #################################################################################################
    # EXERCISE 2: Determine the minimal cost (i.e., the cost of the repair(s) with the least cost). #
    #             Store the minimal cost in variable "min_cost"                                     #
    #################################################################################################
    
    # Return only tables with minimum cost
    cheapest_tables = [(table, cost) for table, cost in repaired_tables_with_cost if cost == min_cost]
    
    return cheapest_tables

# Apply the repair generation process
print("Generating all possible repaired tables...")
repaired_tables_with_cost = apply_repairs(data, violations, verbose=True)
print(f"Number of possible repaired tables: {len(repaired_tables_with_cost)}")

print("\nFiltering tables that satisfy the functional dependency...")
valid_repaired_tables_with_cost = filter_valid_repairs(repaired_tables_with_cost)
print(f"Number of repaired tables that satisfy the functional dependency: {len(valid_repaired_tables_with_cost)}")

print("\nDetermining cheapest repairs...")
cheapest_repaired_tables = determine_cheapest_repairs(valid_repaired_tables_with_cost)
if cheapest_repaired_tables:
    min_cost = cheapest_repaired_tables[0][1]  # All have the same cost
    print(f"Least cost: {min_cost}")
    print(f"Number of repaired tables with least cost: {len(cheapest_repaired_tables)}")
    
    # Display cheapest repaired tables if there are fewer than 4
    if len(cheapest_repaired_tables) < 4:
        print(f"\nDisplaying all {len(cheapest_repaired_tables)} cheapest repaired tables:")
        for i, (table, cost) in enumerate(cheapest_repaired_tables):
            print(f"\nCheapest table {i+1} (cost: {cost}):")
            # Show only relevant columns for the functional dependency
            print(table[['song_title', 'artist', 'album', 'year', 'genre']].head(20))
else:
    print("No valid repaired tables found.")
```

## Optional Exercise 2: Consistent Query Answering (CQA)

### Background

When a database contains violations of integrity constraints (like functional dependencies), different repair strategies can lead to different "correct" versions of the data. **Consistent Query Answering (CQA)** is a principled approach to querying inconsistent databases that returns only answers that would be consistently returned across **all possible valid repairs**.

The key idea behind CQA is:
- Given an inconsistent database and a query
- Generate all possible valid repairs of the database
- Execute the query on each repaired database
- Return only the results that appear in **every single** repaired database

This intersection of results represents the **certain answers** - data that we can be confident about regardless of which specific repair was applied.

### Why is CQA Important?

1. **Uncertainty handling**: When data quality issues exist, CQA provides a conservative but reliable approach
2. **Avoiding arbitrary choices**: Rather than picking one specific repair, CQA considers all reasonable repairs
3. **Decision support**: In critical applications, knowing what is "certainly true" is more valuable than potentially incorrect specific answers

### Example Intuition

If we have violations in our music database:
- Some repairs might delete "Beat It" by Michael Jackson
- Other repairs might keep "Beat It" but change its genre
- A query asking for "all artists" would return different results depending on the repair
- CQA would only return artists that appear in **every possible repair**

This ensures we never return information that might be wrong due to data quality issues.
